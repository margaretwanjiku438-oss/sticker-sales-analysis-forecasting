```python
train_ts = daily_sales[daily_sales.index.year < 2016]
valid_ts = daily_sales[daily_sales.index.year == 2016]

train_ts.shape, valid_ts.shape
```

```python
monthly_baseline = train_ts.groupby(train_ts.index.month).mean()
```

```python
monthly_baseline
```

```python
train_ts = daily_sales[daily_sales.index.year < 2016]
valid_ts = daily_sales[daily_sales.index.year == 2016]

train_ts.shape, valid_ts.shape
```

```python
predictions = valid_ts.index.to_series().map(
    lambda x: monthly_baseline[x.month]
)

predictions.head()
```

```python
predictions.shape
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(valid_ts, predictions)

mae
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

rmse = np.sqrt(mean_squared_error(valid_ts, predictions))

rmse
```

```python
from sklearn.metrics import mean_absolute_percentage_error

mape = mean_absolute_percentage_error(valid_ts, predictions)

mape * 100
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 5))

plt.plot(valid_ts.index, valid_ts.values, label="Actual Sales")
plt.plot(predictions.index, predictions.values, label="Predicted Sales")

plt.title("Actual vs Predicted Daily Sales – 2016")
plt.xlabel("Date")
plt.ylabel("Total Units Sold")
plt.legend()
plt.tight_layout()
plt.show()
```
